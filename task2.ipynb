```python
import pandas as pd
import plotly.express as px
from datetime import datetime
import pytz

# Load data
apps_df = pd.read_csv('Play Store Data.csv')

# Clean the "Installs" column
apps_df = apps_df[~apps_df['Installs'].str.contains('Free', na=False)]  # Remove rows where "Installs" contains 'Free'
apps_df['Installs'] = apps_df['Installs'].str.replace('[+,]', '', regex=True).astype(float)

# Filter top 5 categories based on installs
top_categories = apps_df.groupby('Category')['Installs'].sum().nlargest(5).index
filtered_df = apps_df[apps_df['Category'].isin(top_categories)]

# Exclude categories starting with "A," "C," "G," "S"
excluded_starts = ('A', 'C', 'G', 'S')
filtered_df = filtered_df[~filtered_df['Category'].str.startswith(excluded_starts)]

# Filter installs greater than 1 million
filtered_df = filtered_df[filtered_df['Installs'] > 1_000_000]

# Example country aggregation (replace with real country data if available)
filtered_df['Country'] = 'Global'  # Placeholder; replace with actual country data
category_country_installs = filtered_df.groupby(['Country', 'Category'])['Installs'].sum().reset_index()

# Get current time in IST
ist = pytz.timezone('Asia/Kolkata')
current_time = datetime.now(ist).time()

# Define allowed time range
start_time = datetime.strptime("18:00:00", "%H:%M:%S").time()
end_time = datetime.strptime("20:00:00", "%H:%M:%S").time()

# Check if current time is within allowed range
if start_time <= current_time <= end_time:
    # Create the interactive choropleth map
    fig = px.choropleth(
        category_country_installs,
        locations="Country",
        locationmode="country names",
        color="Installs",
        hover_name="Category",
        title="Global Installs by Categories",
        color_continuous_scale=px.colors.sequential.Plasma
    )
    fig.show()
else:
    print("This graph is available only between 6 PM IST and 8 PM IST.")
```
